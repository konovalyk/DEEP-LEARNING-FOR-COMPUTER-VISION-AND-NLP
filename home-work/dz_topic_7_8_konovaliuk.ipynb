Домашнє завдання до модуля «Практичні приклади використання згорткових мереж»



Вітаємо! Як настрiй? Сподiваємося, що ви з нетерпiнням очікуєте на новий челендж 😉

Сьогодні ваше завдання — розробити та навчити модель глибокого навчання для детекції та класифікації об'єктів у приміщенні, використовуючи набір даних Indoor Objects Detection. Цей набір даних містить зображення з 10 класами об'єктів, які часто зустрічаються в приміщеннях: двері, відчинені двері, дверцята шафи, дверцята холодильника, вікно, стілець, стіл, шафа, диван/кушетка та стовп.

Вам потрібно:

1. Ознайомитися з наданим набором даних та його форматом (YOLO).

2. Обрати відповідну архітектуру нейронної мережі для детекції об'єктів.

3. Навчити модель на наданому наборі даних.

4. Оцінити ефективність моделі та проаналізувати результати.

5. Запропонувати можливі покращення та оптимізації.



Основна мета цього проєкту — розробити модель, яка здатна виявляти та розпізнавати хоча б одну третину з представлених у наборі даних об'єктів у приміщенні.

Важливо, щоб система могла надійно визначати принаймні найбільш поширені предмети, такі як двері, стільці та столи, що дозволить створити базову функціональність для подальшого вдосконалення.

Це завдання дозволить вам застосувати вивчені методи глибокого навчання на практиці та створити робочий прототип системи комп'ютерного зору.



Покрокова інструкція виконання

1. Здійсніть імпорт необхідних бібліотек.

2. Вибір робочого середовища:

Для виконання завдання оберіть між Kaggle Notebooks та Google Colab. Обидві платформи надають безкоштовний доступ до GPU, що суттєво прискорить процес навчання моделі.
Або ж виконуйте завдання на власній машині з використанням NVIDIA GPU.

3. Підготовка даних:

Завантажте набір даних Indoor Objects Detection з платформи Kaggle.
Дані вже розмічено на навчальні, валідаційні й тестові.
4. Знайомство з даними:

Дослідіть дані, як це було зроблено у темі «Детекція об'єктів: огляд архітектур».
5. Конфігурація YOLO:

Створіть файл конфігурації YAML з інформацією про шляхи до даних та кількість класів.
Виберіть відповідну версію YOLO (наприклад, YOLOv9s).
6. Навчання моделі:

Запустіть скрипт навчання YOLO з відповідними параметрами.
Слідкуйте за процесом навчання та метриками.
Збережіть ваги натренованої моделі.
7. Аналіз результатів:

• Проаналізуйте графіки навчання (learning curves) для тренувального та валідаційного наборів даних:

Порівняйте криві втрат (loss curves) для виявлення перенавчання (overfitting) або недонавчання (underfitting).
Оцініть динаміку зміни метрик mAP (mean Average Precision) та інших релевантних показників.
• Проведіть детальний аналіз продуктивності моделі для кожного класу об'єктів:

Порівняйте точність (precision), повноту (recall) та F1-score для кожного класу.
Визначте класи, з якими модель справляється найкраще і найгірше.
• Створіть матрицю помилок (confusion matrix) для оцінки міжкласових помилок:

Виявіть пари класів, які модель часто плутає між собою.
Проаналізуйте можливі причини цих помилок (наприклад, візуальна схожість об'єктів).
• Сформулюйте висновки та рекомендації:

На основі аналізу запропонуйте конкретні кроки для покращення моделі.
Визначте, чи досягнуто поставленої мети щодо точності детекції об'єктів.
Цей глибокий аналіз допоможе вам краще зрозуміти поведінку моделі, виявити її сильні та слабкі сторони, а також визначити напрямки для подальшого вдосконалення.





Орієнтовний очікуваний результат

Орієнтовні результати метрик можуть бути наступними:

1. Mean Average Precision (mAP):

загальний mAP@0.5 (тобто при порозі IoU 0.5): 40-60%,
mAP@0.5:0.95 (усереднений по різних порогах IoU): 25-40%.
2. Точність (Precision) для окремих класів:

для найпростіших об'єктів (наприклад, двері, вікна): 60-75%,
для складніших об'єктів (наприклад, стільці, столи): 50-65%,
для найскладніших об'єктів (наприклад, відкриті двері, дверцята шафи): 40-55%.
3. Повнота (Recall) для окремих класів:

може варіюватися від 50% до 70% залежно від класу.
4. F1-score:

у середньому по всіх класах: 45-60%.
5. Загальна точність класифікації:

55-70%.
☝ Важливо зазначити, що ці показники є приблизними і можуть суттєво відрізнятися.

Студент, який вперше працює з такою задачею, може отримати результати в нижній частині цих діапазонів. З покращенням навичок та застосуванням більш складних технік, результати можуть наближатися до верхньої межі зазначених діапазонів або навіть перевищувати їх.


Ці метрики дають орієнтир для оцінки успішності виконання завдання, але головна мета — отримати працюючу модель, яка може виявляти об'єкти з прийнятною точністю.

Навіть якщо ваша модель не досягла очікуваних метрик, ви все ще можете отримати залік, якщо:

1. Детально проаналізуєте отримані результати:

Поясніть, чому модель могла не досягти очікуваних показників.
Обговоріть можливі обмеження даних або обраного підходу.
2. Критично оціните архітектуру вашої моделі:

Опишіть, чому ви обрали саме таку структуру нейронної мережі.
Розгляньте потенційні недоліки вашої архітектури.
3. Запропонуєте шляхи покращення:

Надайте конкретні ідеї щодо оптимізації моделі.
Обговоріть альтернативні підходи, які могли б дати кращі результати.
☝🏻 Пам'ятайте, що в реальному світі не завжди вдається досягти ідеальних результатів з першої спроби. Вміння аналізувати, робити висновки та пропонувати покращення є надзвичайно цінними навичками в галузі машинного навчання. 
Ваша здатність критично мислити, аналізувати результати та пропонувати обґрунтовані покращення є ключовими факторами для високої оцінки цього завдання.
